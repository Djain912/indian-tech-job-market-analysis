```python
# =============================================================================
# FINAL MINI-PROJECT BACKEND
# AI-ASSISTED INDIAN TECH JOB MARKET ANALYSIS - 2026
# =============================================================================
#
# MAIN PROBLEM:
#
# Given a candidate profile, determine:
#   1. Which jobs are eligible
#   2. Candidate-job compatibility score
#   3. Best matching jobs
#   4. Why a candidate is/isn't eligible
#   5. Missing skills
#   6. Which skills should be learned
#   7. What-if effect of adding skills
#   8. AI explanation using LangChain + Gemini
#   9. Multi-step workflow using LangGraph
#
# IMPORTANT:
#
# - Uses ALL job postings for the candidate matching engine.
# - No salary filtering for candidate-job matching.
# - Generic/noisy terms do not contribute to core skill matching.
# - Match Score is NOT probability of getting hired.
# - The dataset contains no selected/rejected candidate outcomes.
# - Therefore, no fake hiring probability is produced.
#
# ONLY USER INPUTS:
#   1. Upload CSV
#   2. Gemini API key
#
# Candidate information below is fixed demonstration data.
# Future Streamlit UI will replace the demo values.
# =============================================================================


# =============================================================================
# 1. INSTALL PACKAGES
# =============================================================================

import sys
import subprocess
import importlib.util


packages = {
    "pandas": "pandas",
    "numpy": "numpy",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "joblib": "joblib",
    "langchain": "langchain",
    "langgraph": "langgraph",
    "langchain-google-genai": "langchain_google_genai"
}


for package_name, module_name in packages.items():

    if importlib.util.find_spec(
        module_name
    ) is None:

        print(
            f"Installing {package_name}..."
        )

        subprocess.check_call(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                package_name
            ]
        )


# =============================================================================
# 2. IMPORTS
# =============================================================================

import os
import re
import json
import warnings
from pathlib import Path
from typing import TypedDict, List, Dict, Any, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from IPython.display import display

warnings.filterwarnings("ignore")


print("=" * 80)
print("AI-ASSISTED INDIAN TECH JOB MARKET ANALYSIS - 2026")
print("=" * 80)


# =============================================================================
# 3. CSV UPLOAD
# =============================================================================

from google.colab import files

print("\nUPLOAD YOUR JOB MARKET CSV")
print("-" * 80)

uploaded = files.upload()


csv_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith(".csv")
]


if not csv_files:

    raise ValueError(
        "No CSV file was uploaded."
    )


if len(csv_files) == 1:

    CSV_FILE = csv_files[0]

else:

    print(
        "\nMultiple CSV files detected:"
    )

    for i, filename in enumerate(
        csv_files,
        start=1
    ):

        print(
            f"{i}. {filename}"
        )

    file_number = int(
        input(
            "\nEnter CSV number: "
        )
    )

    if not (
        1 <= file_number <= len(csv_files)
    ):

        raise ValueError(
            "Invalid CSV number."
        )

    CSV_FILE = csv_files[
        file_number - 1
    ]


print(
    "\nSelected file:",
    CSV_FILE
)


# =============================================================================
# 4. LOAD DATASET
# =============================================================================

df_raw = pd.read_csv(
    CSV_FILE
)


print("\n" + "=" * 80)
print("DATASET LOADED")
print("=" * 80)

print(
    "Rows:",
    len(df_raw)
)

print(
    "Columns:",
    len(df_raw.columns)
)

display(
    df_raw.head()
)


# =============================================================================
# 5. REQUIRED COLUMNS
# =============================================================================

required_columns = [

    "job_id",
    "job_title",
    "company_name",
    "location",
    "role_category",
    "experience_raw",
    "experience_min_yrs",
    "experience_max_yrs",
    "skills_required",
    "skills_count",
    "work_mode",
    "company_size_bucket",
    "primary_city",
    "skill_domain",
    "salary_midpoint_lpa",
    "is_fresher_friendly"

]


missing_columns = [

    column

    for column
    in required_columns

    if column not in df_raw.columns

]


if missing_columns:

    raise ValueError(
        "Missing required columns: "
        +
        str(
            missing_columns
        )
    )


print(
    "\nAll required columns are present."
)


# =============================================================================
# 6. INITIAL QUALITY CHECK
# =============================================================================

print("\n" + "=" * 80)
print("INITIAL DATA QUALITY CHECK")
print("=" * 80)

print(
    "Duplicate rows:",
    df_raw.duplicated().sum()
)

print(
    "Duplicate job IDs:",
    df_raw[
        "job_id"
    ].duplicated().sum()
)


print("\nMissing values:")

display(

    df_raw
    .isnull()
    .sum()
    .sort_values(
        ascending=False
    )
    .to_frame(
        "missing_count"
    )

)


# =============================================================================
# 7. CLEAN DATASET
# =============================================================================

df = df_raw.copy()


df = df.drop_duplicates()


df = df.drop_duplicates(

    subset=[
        "job_id"
    ],

    keep="first"

)


# =============================================================================
# 8. TEXT CLEANING
# =============================================================================

text_columns = [

    "job_title",
    "company_name",
    "location",
    "scraped_city",
    "role_category",
    "experience_raw",
    "skills_required",
    "job_description",
    "posted_date_raw",
    "work_mode",
    "company_size_bucket",
    "salary_raw",
    "salary_tier",
    "experience_tier",
    "primary_city",
    "skill_domain",
    "data_source",
    "scraped_at"

]


for column in text_columns:

    if column in df.columns:

        df[column] = (

            df[column]
            .astype("string")
            .str.strip()

        )


# =============================================================================
# 9. NUMERIC CLEANING
# =============================================================================

numeric_columns = [

    "company_rating",
    "experience_min_yrs",
    "experience_max_yrs",
    "salary_min_lpa",
    "salary_max_lpa",
    "skills_count",
    "salary_midpoint_lpa",
    "days_since_posted"

]


for column in numeric_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(

            df[column],

            errors="coerce"

        )


# =============================================================================
# 10. BOOLEAN CLEANING
# =============================================================================

def clean_boolean(
    series
):

    if pd.api.types.is_bool_dtype(
        series
    ):

        return (

            series
            .fillna(False)
            .astype(bool)

        )


    mapping = {

        "true": True,
        "false": False,

        "yes": True,
        "no": False,

        "1": True,
        "0": False

    }


    return (

        series
        .astype(str)
        .str.lower()
        .map(mapping)
        .fillna(False)
        .astype(bool)

    )


for column in [

    "salary_disclosed",
    "is_senior",
    "is_fresher_friendly",
    "salary_negotiable"

]:

    if column in df.columns:

        df[column] = clean_boolean(
            df[column]
        )


# =============================================================================
# 11. CLEAN EXPERIENCE
# =============================================================================

def parse_min_experience(

    raw_value,

    existing_value

):

    if pd.isna(
        raw_value
    ):

        if pd.isna(
            existing_value
        ):

            return np.nan

        return float(
            existing_value
        )


    text = (

        str(
            raw_value
        )
        .lower()
        .strip()

    )


    if "fresher" in text:

        return 0.0


    match = re.search(

        r"(\d+(?:\.\d+)?)\s*[-–]\s*(\d+(?:\.\d+)?)",

        text

    )


    if match:

        return float(
            match.group(1)
        )


    match = re.search(

        r"(\d+(?:\.\d+)?)\s*\+",

        text

    )


    if match:

        return float(
            match.group(1)
        )


    match = re.search(

        r"(\d+(?:\.\d+)?)\s*(?:year|years|yr|yrs)",

        text

    )


    if match:

        return float(
            match.group(1)
        )


    if not pd.isna(
        existing_value
    ):

        return float(
            existing_value
        )


    return np.nan


df[
    "clean_experience_min_yrs"
] = [

    parse_min_experience(

        raw,

        existing

    )

    for raw, existing

    in zip(

        df[
            "experience_raw"
        ],

        df[
            "experience_min_yrs"
        ]

    )

]


# =============================================================================
# 12. DERIVED COLUMNS
# =============================================================================

df[
    "salary_available"
] = (

    df[
        "salary_midpoint_lpa"
    ].notna()

    &

    (
        df[
            "salary_midpoint_lpa"
        ]
        >
        0
    )

)


df[
    "remote_or_hybrid"
] = (

    df[
        "work_mode"
    ]
    .astype(str)
    .str.lower()
    .isin(
        [
            "remote",
            "hybrid"
        ]
    )

)


# =============================================================================
# 13. DATA SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("CLEANED DATASET SUMMARY")
print("=" * 80)

print(
    "Original records:",
    len(df_raw)
)

print(
    "Cleaned records:",
    len(df)
)

print(
    "Unique companies:",
    df[
        "company_name"
    ].nunique()
)

print(
    "Unique job titles:",
    df[
        "job_title"
    ].nunique()
)

print(
    "Unique cities:",
    df[
        "primary_city"
    ].nunique()
)

print(
    "Role categories:",
    df[
        "role_category"
    ].nunique()
)

print(
    "Usable salary records:",
    int(
        df[
            "salary_available"
        ].sum()
    )
)

print(
    "Fresher-friendly jobs:",
    int(
        df[
            "is_fresher_friendly"
        ].sum()
    )
)

print(
    "Remote/Hybrid jobs:",
    int(
        df[
            "remote_or_hybrid"
        ].sum()
    )
)


# =============================================================================
# 14. MARKET VISUALIZATIONS
# =============================================================================

print("\n" + "=" * 80)
print("MARKET VISUALIZATIONS")
print("=" * 80)


# ---------------------------------------
# Role distribution
# ---------------------------------------

role_counts = (

    df[
        "role_category"
    ]
    .value_counts()
    .sort_values()

)


plt.figure(
    figsize=(10, 6)
)

plt.barh(

    role_counts.index,

    role_counts.values

)

plt.title(
    "Job Postings by Role Category"
)

plt.xlabel(
    "Number of Jobs"
)

plt.ylabel(
    "Role"
)

plt.tight_layout()

plt.show()


# ---------------------------------------
# Top cities
# ---------------------------------------

city_counts = (

    df[
        "primary_city"
    ]
    .value_counts()
    .head(15)
    .sort_values()

)


plt.figure(
    figsize=(10, 6)
)

plt.barh(

    city_counts.index,

    city_counts.values

)

plt.title(
    "Top 15 Cities by Job Postings"
)

plt.xlabel(
    "Number of Jobs"
)

plt.ylabel(
    "City"
)

plt.tight_layout()

plt.show()


# ---------------------------------------
# Work mode
# ---------------------------------------

work_counts = (

    df[
        "work_mode"
    ]
    .value_counts()

)


plt.figure(
    figsize=(8, 5)
)

plt.bar(

    work_counts.index.astype(str),

    work_counts.values

)

plt.title(
    "Job Distribution by Work Mode"
)

plt.xlabel(
    "Work Mode"
)

plt.ylabel(
    "Number of Jobs"
)

plt.tight_layout()

plt.show()


# ---------------------------------------
# Experience
# ---------------------------------------

experience_counts = (

    df[
        "experience_tier"
    ]
    .value_counts()

)


plt.figure(
    figsize=(10, 6)
)

plt.bar(

    experience_counts.index.astype(str),

    experience_counts.values

)

plt.title(
    "Job Distribution by Experience Tier"
)

plt.xlabel(
    "Experience Tier"
)

plt.ylabel(
    "Number of Jobs"
)

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()


# ---------------------------------------
# Fresher opportunities
# ---------------------------------------

fresher_role = (

    df

    .groupby(
        "role_category"
    )[

        "is_fresher_friendly"

    ]

    .mean()

    .sort_values()

    *
    100

)


plt.figure(
    figsize=(10, 6)
)

plt.barh(

    fresher_role.index,

    fresher_role.values

)

plt.title(
    "Fresher-Friendly Jobs by Role"
)

plt.xlabel(
    "Fresher-Friendly Percentage"
)

plt.ylabel(
    "Role"
)

plt.tight_layout()

plt.show()


# =============================================================================
# 15. SALARY ANALYSIS
# =============================================================================

print("\n" + "=" * 80)
print("SALARY ANALYSIS")
print("=" * 80)


salary_df = df[
    df[
        "salary_available"
    ]
].copy()


print(
    "Total jobs:",
    len(df)
)

print(
    "Usable salary records:",
    len(salary_df)
)


if len(
    salary_df
) > 0:

    print(
        "Average disclosed salary:",
        round(
            float(
                salary_df[
                    "salary_midpoint_lpa"
                ].mean()
            ),
            2
        ),
        "LPA"
    )


    print(
        "\nSalary by role:"
    )


    salary_by_role = (

        salary_df

        .groupby(
            "role_category"
        )[

            "salary_midpoint_lpa"

        ]

        .agg(

            [
                "count",
                "mean",
                "median"

            ]

        )

        .sort_values(

            "mean",

            ascending=False

        )

    )


    display(
        salary_by_role.round(2)
    )


# =============================================================================
# 16. SKILL TAXONOMY
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING SKILL TAXONOMY")
print("=" * 80)


# -------------------------------------------------------------------------
# CORE TECHNICAL SKILLS
# -------------------------------------------------------------------------

technical_skills = {

    "python",
    "java",
    "javascript",
    "typescript",
    "c",
    "c++",
    "c#",
    "r",
    "sql",
    "html",
    "css",

    "excel",
    "power bi",
    "tableau",

    "data analysis",
    "data visualization",
    "data modeling",
    "data management",
    "data quality",
    "business intelligence",
    "statistics",

    "data science",
    "machine learning",
    "deep learning",
    "artificial intelligence",
    "natural language processing",
    "nlp",
    "generative ai",
    "gen ai",
    "llm",
    "large language model",
    "rag",
    "retrieval augmented generation",
    "agentic ai",
    "predictive modeling",

    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "pytorch",

    "data engineering",
    "spark",
    "pyspark",
    "kafka",
    "airflow",

    "aws",
    "azure",
    "gcp",
    "cloud",

    "docker",
    "kubernetes",
    "git",

    "mlops",
    "ci/cd",

    "automation"

}


# -------------------------------------------------------------------------
# BUSINESS SKILLS
# -------------------------------------------------------------------------

business_skills = {

    "business analysis",
    "business analyst",
    "project management",
    "agile",
    "consulting",
    "change management",
    "market research",
    "stakeholder management",
    "erp",
    "sap"

}


# -------------------------------------------------------------------------
# GENERIC TERMS
# -------------------------------------------------------------------------

generic_skills = {

    "data",
    "analysis",
    "analytical",
    "analytics",
    "coding",
    "basic",
    "computer science",
    "management",
    "communication",
    "leadership",
    "teamwork",
    "planning",
    "support",
    "operations",
    "reporting",
    "technical",
    "professional",
    "knowledge",
    "experience",
    "research",
    "development",
    "skills",
    "responsibilities"

}


# -------------------------------------------------------------------------
# ALIAS MAPPING
# -------------------------------------------------------------------------

skill_aliases = {

    "python programming":
        "python",

    "python 3":
        "python",

    "sql server":
        "sql",

    "structured query language":
        "sql",

    "ml":
        "machine learning",

    "machine-learning":
        "machine learning",

    "ai":
        "artificial intelligence",

    "powerbi":
        "power bi",

    "power-bi":
        "power bi",

    "ms power bi":
        "power bi",

    "ms excel":
        "excel",

    "microsoft excel":
        "excel",

    "genai":
        "generative ai",

    "gen-ai":
        "generative ai",

    "llms":
        "llm",

    "scikit learn":
        "scikit-learn"

}


# =============================================================================
# 17. SKILL NORMALIZATION
# =============================================================================

def normalize_skill_name(
    skill
):

    if skill is None:

        return ""


    try:

        if pd.isna(
            skill
        ):

            return ""

    except (
        TypeError,
        ValueError
    ):

        pass


    skill = (

        str(
            skill
        )
        .lower()
        .strip()

    )


    if skill in {

        "",
        "nan",
        "none",
        "<na>",
        "na",
        "n/a",
        "not available",
        "not applicable"

    }:

        return ""


    skill = (
        skill_aliases
        .get(
            skill,
            skill
        )
    )


    skill = re.sub(

        r"\s+",
        " ",
        skill

    )


    return skill


def normalize_skill_list(
    skills
):

    if skills is None:

        return set()


    if isinstance(
        skills,
        str
    ):

        if not skills.strip():

            return set()


        skills = re.split(

            r"[,;|]",

            skills

        )


    elif not isinstance(

        skills,

        (
            list,
            tuple,
            set,
            np.ndarray

        )

    ):

        try:

            if pd.isna(
                skills
            ):

                return set()

        except (
            TypeError,
            ValueError
        ):

            pass


        skills = [
            skills
        ]


    result = set()


    for skill in skills:

        normalized = (
            normalize_skill_name(
                skill
            )
        )


        if normalized:

            result.add(
                normalized
            )


    return result


def skill_weight(
    skill
):

    normalized = (
        normalize_skill_name(
            skill
        )
    )


    if normalized in technical_skills:

        return 1.0


    if normalized in business_skills:

        return 0.5


    return 0.0


# =============================================================================
# 18. CREATE SKILL DATASET
# =============================================================================

skill_source = df[
    [
        "job_id",
        "role_category",
        "skills_required",
        "primary_city",
        "work_mode",
        "experience_tier",
        "salary_midpoint_lpa",
        "salary_available",
        "is_fresher_friendly"

    ]
].copy()


skill_source[
    "skills_required"
] = (

    skill_source[
        "skills_required"
    ]
    .fillna("")
    .astype(str)

)


skill_long = (

    skill_source

    .assign(

        skill=

        skill_source[
            "skills_required"
        ]

        .str.split(
            r"[,;|]"
        )

    )

    .explode(
        "skill"
    )

)


skill_long[
    "skill"
] = (

    skill_long[
        "skill"
    ]

    .apply(
        normalize_skill_name
    )

)


skill_long = skill_long[
    skill_long[
        "skill"
    ] != ""
].copy()


skill_long[
    "skill_category"
] = (

    skill_long[
        "skill"
    ]
    .apply(

        lambda x:

        (
            "technical"
            if x in technical_skills

            else

            "business"
            if x in business_skills

            else

            "generic"
            if x in generic_skills

            else

            "other"
        )

    )

)


skill_long[
    "skill_weight"
] = (

    skill_long[
        "skill"
    ]
    .apply(
        skill_weight
    )

)


skill_long = (

    skill_long
    .drop_duplicates(

        subset=[
            "job_id",
            "skill"
        ]

    )

)


print(
    "Skill-job records:",
    len(skill_long)
)


print(
    "\nSkill categories:"
)


display(

    skill_long[
        "skill_category"
    ]
    .value_counts()
    .to_frame(
        "count"
    )

)


# =============================================================================
# 19. TOP MARKET SKILLS
# =============================================================================

top_skills = (

    skill_long

    .groupby(
        "skill"
    )[

        "job_id"

    ]

    .nunique()

    .sort_values(
        ascending=False
    )

    .head(20)

)


print(
    "\nTop 20 market skills:"
)


display(

    pd.DataFrame({

        "jobs":
            top_skills,

        "demand_percent":
            (
                top_skills
                /
                len(df)
                *
                100
            ).round(2)

    })

)


# =============================================================================
# 20. ROLE SKILL ANALYSIS
# =============================================================================

target_role_for_analysis = (
    "Data Analyst"
)


role_core_skills = (

    skill_long[

        (

            skill_long[
                "role_category"
            ]
            .astype(str)
            .str.lower()
            ==
            target_role_for_analysis.lower()

        )

        &

        (

            skill_long[
                "skill_weight"
            ]
            >
            0

        )

    ]

    .groupby(
        "skill"
    )[

        "job_id"

    ]

    .nunique()

    .sort_values(
        ascending=False
    )

    .head(15)

)


print(
    f"\nTop core skills for {target_role_for_analysis}:"
)


display(
    role_core_skills.to_frame(
        "job_count"
    )
)


# =============================================================================
# 21. ROLE OPPORTUNITY SCORE
# =============================================================================

print("\n" + "=" * 80)
print("ROLE OPPORTUNITY SCORE")
print("=" * 80)


def minmax_scale(
    series
):

    series = (
        series
        .astype(float)
    )


    if (
        series.max()
        ==
        series.min()
    ):

        return pd.Series(

            1.0,

            index=series.index

        )


    return (

        (
            series
            -
            series.min()
        )

        /

        (
            series.max()
            -
            series.min()
        )

    )


role_analysis = (

    df

    .groupby(
        "role_category"
    )

    .agg(

        job_count=(
            "job_id",
            "count"
        ),

        fresher_rate=(
            "is_fresher_friendly",
            "mean"
        ),

        flexibility_rate=(
            "remote_or_hybrid",
            "mean"
        )

    )

)


if len(
    salary_df
) > 0:

    role_salary = (

        salary_df

        .groupby(
            "role_category"
        )[

            "salary_midpoint_lpa"

        ]

        .mean()

        .rename(
            "avg_salary"
        )

    )


    role_analysis = (

        role_analysis
        .join(
            role_salary,
            how="left"
        )

    )


    role_analysis[
        "avg_salary"
    ] = (

        role_analysis[
            "avg_salary"
        ]

        .fillna(

            salary_df[
                "salary_midpoint_lpa"
            ].median()

        )

    )

else:

    role_analysis[
        "avg_salary"
    ] = 0


role_analysis[
    "demand_score"
] = minmax_scale(

    role_analysis[
        "job_count"
    ]

)


role_analysis[
    "salary_score"
] = minmax_scale(

    role_analysis[
        "avg_salary"
    ]

)


role_analysis[
    "fresher_score"
] = (

    role_analysis[
        "fresher_rate"
    ]
    .clip(
        0,
        1
    )

)


role_analysis[
    "flexibility_score"
] = (

    role_analysis[
        "flexibility_rate"
    ]
    .clip(
        0,
        1
    )

)


role_analysis[
    "opportunity_score_100"
] = (

    (

        0.45
        *
        role_analysis[
            "demand_score"
        ]

    )

    +

    (

        0.25
        *
        role_analysis[
            "salary_score"
        ]

    )

    +

    (

        0.20
        *
        role_analysis[
            "fresher_score"
        ]

    )

    +

    (

        0.10
        *
        role_analysis[
            "flexibility_score"
        ]

    )

) * 100


role_analysis = (

    role_analysis

    .sort_values(

        "opportunity_score_100",

        ascending=False

    )

)


display(

    role_analysis[

        [
            "job_count",
            "avg_salary",
            "fresher_rate",
            "flexibility_rate",
            "opportunity_score_100"

        ]

    ].round(2)

)


# =============================================================================
# 22. JOB SKILL CACHE
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING JOB SKILL CACHE")
print("=" * 80)


job_skill_cache = {}


for job_id, skills_text in zip(

    df[
        "job_id"
    ],

    df[
        "skills_required"
    ]

):

    all_skills = (
        normalize_skill_list(
            skills_text
        )
    )


    # Only core skills used for matching.
    core_skills = {

        skill

        for skill
        in all_skills

        if skill_weight(
            skill
        ) > 0

    }


    job_skill_cache[
        job_id
    ] = core_skills


print(
    "Total jobs:",
    len(df)
)

print(
    "Cached jobs:",
    len(job_skill_cache)
)


# =============================================================================
# 23. ROLE-TITLE RELEVANCE
# =============================================================================

ROLE_TITLE_KEYWORDS = {

    "Data Analyst": [

        "data analyst",
        "data analysis",
        "data analytics",
        "analytics analyst",
        "business intelligence analyst",
        "bi analyst",
        "reporting analyst",
        "mis analyst",
        "data reporting",
        "data insights",
        "data visualization analyst"

    ],

    "Data Scientist": [

        "data scientist",
        "data science",
        "applied scientist",
        "research scientist"

    ],

    "Data Engineer": [

        "data engineer",
        "data engineering",
        "big data engineer",
        "etl developer",
        "etl engineer",
        "data pipeline engineer",
        "data platform engineer"

    ],

    "Machine Learning Engineer": [

        "machine learning engineer",
        "ml engineer",
        "machine learning",
        "ai engineer",
        "artificial intelligence engineer",
        "deep learning engineer"

    ],

    "Business Analyst": [

        "business analyst",
        "business analysis",
        "business systems analyst",
        "functional analyst",
        "requirements analyst"

    ],

    "Python Developer": [

        "python developer",
        "python programmer",
        "python engineer",
        "django developer",
        "flask developer",
        "python backend"

    ]

}


def role_title_relevance(

    job_title,

    target_role

):

    if not target_role:

        return 50.0


    title = (

        str(
            job_title
        )
        .lower()
        .strip()

    )


    keywords = (
        ROLE_TITLE_KEYWORDS
        .get(
            target_role,
            []
        )
    )


    for keyword in keywords:

        if keyword in title:

            return 100.0


    return 0.0


# =============================================================================
# 24. CITY NORMALIZATION
# =============================================================================

city_aliases = {

    "bangalore": [

        "bangalore",
        "bengaluru"

    ],

    "mumbai": [

        "mumbai",
        "bombay"

    ],

    "kolkata": [

        "kolkata",
        "calcutta"

    ],

    "chennai": [

        "chennai",
        "madras"

    ],

    "gurgaon": [

        "gurgaon",
        "gurugram"

    ]

}


def normalize_city(
    city
):

    if not city:

        return ""


    city = (

        str(
            city
        )
        .lower()
        .strip()

    )


    for canonical, aliases in (

        city_aliases.items()

    ):

        if city in aliases:

            return canonical


    return city


# =============================================================================
# 25. WORK MODE NORMALIZATION
# =============================================================================

def normalize_work_modes(
    value
):

    if not value:

        return []


    if isinstance(

        value,

        (
            list,
            tuple,
            set
        )

    ):

        values = list(
            value
        )

    else:

        values = re.split(

            r",|/|\bor\b",

            str(value),

            flags=re.IGNORECASE

        )


    result = []


    for value in values:

        mode = (

            str(
                value
            )
            .lower()
            .strip()

        )


        if mode in [

            "onsite",
            "on site"

        ]:

            mode = "on-site"


        if mode in [

            "remote",
            "hybrid",
            "on-site"

        ]:

            if mode not in result:

                result.append(
                    mode
                )


    return result


# =============================================================================
# 26. SKILL MATCHING
# =============================================================================

def calculate_skill_match(

    candidate_skills,

    required_skills

):

    candidate_set = (
        normalize_skill_list(
            candidate_skills
        )
    )


    candidate_set = {

        skill

        for skill
        in candidate_set

        if skill_weight(
            skill
        ) > 0

    }


    required_set = {

        skill

        for skill
        in required_skills

        if skill_weight(
            skill
        ) > 0

    }


    if not required_set:

        return {

            "score":
                0.0,

            "matched":
                [],

            "missing":
                [],

            "required_count":
                0

        }


    matched = (

        candidate_set
        &
        required_set

    )


    missing = (

        required_set
        -
        candidate_set

    )


    total_weight = sum(

        skill_weight(
            skill
        )

        for skill
        in required_set

    )


    matched_weight = sum(

        skill_weight(
            skill
        )

        for skill
        in matched

    )


    score = (

        matched_weight
        /
        total_weight
        *
        100

    )


    return {

        "score":
            round(
                float(
                    score
                ),
                2
            ),

        "matched":
            sorted(
                matched
            ),

        "missing":
            sorted(
                missing
            ),

        "required_count":
            len(
                required_set
            )

    }


# =============================================================================
# 27. CANDIDATE-JOB ANALYSIS
# =============================================================================

def analyze_job_for_candidate(

    job_row,

    candidate_skills,

    candidate_experience,

    preferred_city=None,

    preferred_work_mode=None,

    target_role=None

):

    job_id = job_row[
        "job_id"
    ]


    # -------------------------------------------------------------------------
    # SKILLS
    # -------------------------------------------------------------------------

    required_skills = (
        job_skill_cache.get(
            job_id,
            set()
        )
    )


    skill_result = (
        calculate_skill_match(

            candidate_skills,

            required_skills

        )
    )


    skill_score = (
        skill_result[
            "score"
        ]
    )


    # -------------------------------------------------------------------------
    # ROLE RELEVANCE
    # -------------------------------------------------------------------------

    role_relevance = (
        role_title_relevance(

            job_row[
                "job_title"
            ],

            target_role

        )
    )


    # -------------------------------------------------------------------------
    # EXPERIENCE
    # -------------------------------------------------------------------------

    minimum_experience = (

        job_row[
            "clean_experience_min_yrs"
        ]

    )


    if pd.isna(
        minimum_experience
    ):

        experience_score = 50.0

        experience_eligible = True

        experience_reason = (
            "No reliable minimum experience requirement was available."
        )

    elif candidate_experience >= minimum_experience:

        experience_score = 100.0

        experience_eligible = True

        experience_reason = (

            f"Experience requirement satisfied: "
            f"{candidate_experience:g} years candidate; "
            f"{minimum_experience:g} years required."

        )

    else:

        experience_score = (

            candidate_experience
            /
            max(
                minimum_experience,
                1
            )
            *
            100

        )


        experience_score = max(

            0.0,

            min(
                100.0,
                experience_score
            )

        )


        experience_eligible = False

        experience_reason = (

            f"Candidate has {candidate_experience:g} years "
            f"but the job requires at least "
            f"{minimum_experience:g} years."

        )


    # -------------------------------------------------------------------------
    # LOCATION
    # -------------------------------------------------------------------------

    if not preferred_city:

        location_score = 50.0

        location_reason = (
            "No city preference."
        )

    else:

        candidate_city = normalize_city(
            preferred_city
        )


        job_location = (

            str(
                job_row[
                    "location"
                ]
            )
            .lower()

        )


        city_alias_list = (

            city_aliases.get(

                candidate_city,

                [candidate_city]

            )

        )


        if any(

            alias
            in
            job_location

            for alias
            in
            city_alias_list

        ):

            location_score = 100.0

            location_reason = (
                "Preferred city is compatible with job location."
            )

        else:

            location_score = 25.0

            location_reason = (
                "Preferred city does not match job location."
            )


    # -------------------------------------------------------------------------
    # WORK MODE
    # -------------------------------------------------------------------------

    preferred_modes = (
        normalize_work_modes(
            preferred_work_mode
        )
    )


    job_mode = (

        str(
            job_row[
                "work_mode"
            ]
        )
        .lower()
        .strip()

    )


    if job_mode in [
        "onsite",
        "on site"
    ]:

        job_mode = "on-site"


    if not preferred_modes:

        work_mode_score = 50.0

        work_mode_reason = (
            "No work-mode preference."
        )

    elif job_mode in preferred_modes:

        work_mode_score = 100.0

        work_mode_reason = (
            "Job work mode matches preference."
        )

    elif (

        "remote"
        in
        preferred_modes

        and

        job_mode
        ==
        "hybrid"

    ):

        work_mode_score = 75.0

        work_mode_reason = (
            "Hybrid is partially compatible with remote preference."
        )

    elif (

        "hybrid"
        in
        preferred_modes

        and

        job_mode
        ==
        "remote"

    ):

        work_mode_score = 75.0

        work_mode_reason = (
            "Remote is partially compatible with hybrid preference."
        )

    else:

        work_mode_score = 25.0

        work_mode_reason = (
            "Job work mode does not match preference."
        )


    # -------------------------------------------------------------------------
    # MATCH SCORE
    #
    # 50% Skills
    # 25% Experience
    # 10% Role Relevance
    # 10% Location
    # 5% Work Mode
    # -------------------------------------------------------------------------

    match_score = (

        0.50
        *
        skill_score

        +

        0.25
        *
        experience_score

        +

        0.10
        *
        role_relevance

        +

        0.10
        *
        location_score

        +

        0.05
        *
        work_mode_score

    )


    # -------------------------------------------------------------------------
    # ELIGIBILITY
    #
    # Requires:
    # 1. Minimum experience satisfied
    # 2. At least 50% core skill match
    # 3. Relevant job title for target role
    # -------------------------------------------------------------------------

    skill_eligible = (
        skill_score >= 50.0
    )


    role_eligible = (

        target_role is None

        or

        role_relevance >= 100.0

    )


    eligible = (

        skill_eligible

        and

        experience_eligible

        and

        role_eligible

    )


    # -------------------------------------------------------------------------
    # FIT CATEGORY
    # -------------------------------------------------------------------------

    if match_score >= 80:

        fit_category = (
            "Excellent Fit"
        )

    elif match_score >= 65:

        fit_category = (
            "Good Fit"
        )

    elif match_score >= 50:

        fit_category = (
            "Potential Fit"
        )

    elif match_score >= 30:

        fit_category = (
            "Low Fit"
        )

    else:

        fit_category = (
            "Very Low Fit"
        )


    # -------------------------------------------------------------------------
    # APPLICATION PRIORITY
    # -------------------------------------------------------------------------

    if eligible:

        if match_score >= 80:

            priority = (
                "High Priority"
            )

        elif match_score >= 65:

            priority = (
                "Recommended"
            )

        else:

            priority = (
                "Possible"
            )

    else:

        if match_score >= 50:

            priority = (
                "Improve Before Applying"
            )

        else:

            priority = (
                "Low Priority"
            )


    # -------------------------------------------------------------------------
    # EXPLANATION
    # -------------------------------------------------------------------------

    if eligible:

        explanation = (

            f"Eligible because the candidate has "
            f"{skill_score:.1f}% core skill match, "
            f"meets the experience requirement, and "
            f"the job title is relevant to the target role."

        )

    else:

        failed_reasons = []


        if not skill_eligible:

            failed_reasons.append(

                f"core skill match is only "
                f"{skill_score:.1f}%"

            )


        if not experience_eligible:

            failed_reasons.append(
                experience_reason
            )


        if target_role and not role_eligible:

            failed_reasons.append(

                "the job title is not sufficiently "
                f"relevant to {target_role}"

            )


        explanation = (

            "Not eligible because "

            +

            "; ".join(
                failed_reasons
            )

            +

            "."

        )


    # -------------------------------------------------------------------------
    # SALARY
    # -------------------------------------------------------------------------

    salary_value = None


    if (

        bool(
            job_row[
                "salary_available"
            ]
        )

        and

        not pd.isna(
            job_row[
                "salary_midpoint_lpa"
            ]
        )

    ):

        salary_value = round(

            float(
                job_row[
                    "salary_midpoint_lpa"
                ]
            ),

            2

        )


    # -------------------------------------------------------------------------
    # RETURN
    # -------------------------------------------------------------------------

    return {

        "job_id":
            job_id,

        "job_title":
            job_row[
                "job_title"
            ],

        "company":
            job_row[
                "company_name"
            ],

        "role":
            job_row[
                "role_category"
            ],

        "city":
            job_row[
                "primary_city"
            ],

        "work_mode":
            job_row[
                "work_mode"
            ],

        "required_experience":
            (

                float(
                    minimum_experience
                )

                if not pd.isna(
                    minimum_experience
                )

                else None

            ),

        "candidate_experience":
            float(
                candidate_experience
            ),

        "skill_match_percent":
            round(
                float(
                    skill_score
                ),
                2
            ),

        "experience_match_percent":
            round(
                float(
                    experience_score
                ),
                2
            ),

        "role_relevance_percent":
            round(
                float(
                    role_relevance
                ),
                2
            ),

        "location_match_percent":
            round(
                float(
                    location_score
                ),
                2
            ),

        "work_mode_match_percent":
            round(
                float(
                    work_mode_score
                ),
                2
            ),

        "match_score":
            round(
                float(
                    match_score
                ),
                2
            ),

        "eligible":
            bool(
                eligible
            ),

        "fit_category":
            fit_category,

        "application_priority":
            priority,

        "matched_skills":
            skill_result[
                "matched"
            ],

        "missing_skills":
            skill_result[
                "missing"
            ],

        "required_core_skill_count":
            skill_result[
                "required_count"
            ],

        "eligibility_explanation":
            explanation,

        "salary_lpa":
            salary_value

    }


# =============================================================================
# 28. ANALYZE CANDIDATE AGAINST ALL JOBS
# =============================================================================

def analyze_candidate(

    candidate_skills,

    candidate_experience,

    preferred_city=None,

    preferred_work_mode=None,

    target_role=None

):

    candidate_skills = (
        normalize_skill_list(
            candidate_skills
        )
    )


    results = []


    # IMPORTANT:
    # We scan the COMPLETE dataset here.
    # Target role is evaluated inside the scoring engine
    # instead of reducing the dataset before analysis.

    for _, job_row in df.iterrows():

        result = (

            analyze_job_for_candidate(

                job_row,

                candidate_skills,

                candidate_experience,

                preferred_city,

                preferred_work_mode,

                target_role

            )

        )


        results.append(
            result
        )


    result_df = pd.DataFrame(
        results
    )


    if result_df.empty:

        return result_df


    result_df = (

        result_df

        .sort_values(

            by=[
                "eligible",
                "match_score"
            ],

            ascending=[
                False,
                False
            ]

        )

        .reset_index(
            drop=True
        )

    )


    return result_df


# =============================================================================
# 29. OPPORTUNITY DASHBOARD
# =============================================================================

def create_candidate_dashboard(
    result_df
):

    if result_df.empty:

        return {

            "jobs_analyzed":
                0,

            "eligible_jobs":
                0,

            "not_eligible_jobs":
                0,

            "eligibility_percent":
                0,

            "excellent_fit":
                0,

            "good_fit":
                0,

            "potential_fit":
                0,

            "low_fit":
                0,

            "very_low_fit":
                0,

            "average_match_score":
                0

        }


    total = len(
        result_df
    )


    eligible = int(

        result_df[
            "eligible"
        ].sum()

    )


    return {

        "jobs_analyzed":
            total,

        "eligible_jobs":
            eligible,

        "not_eligible_jobs":
            total - eligible,

        "eligibility_percent":
            round(
                eligible
                /
                total
                *
                100,
                2
            ),

        "excellent_fit":
            int(
                (
                    result_df[
                        "fit_category"
                    ]
                    ==
                    "Excellent Fit"
                ).sum()
            ),

        "good_fit":
            int(
                (
                    result_df[
                        "fit_category"
                    ]
                    ==
                    "Good Fit"
                ).sum()
            ),

        "potential_fit":
            int(
                (
                    result_df[
                        "fit_category"
                    ]
                    ==
                    "Potential Fit"
                ).sum()
            ),

        "low_fit":
            int(
                (
                    result_df[
                        "fit_category"
                    ]
                    ==
                    "Low Fit"
                ).sum()
            ),

        "very_low_fit":
            int(
                (
                    result_df[
                        "fit_category"
                    ]
                    ==
                    "Very Low Fit"
                ).sum()
            ),

        "average_match_score":
            round(
                float(
                    result_df[
                        "match_score"
                    ].mean()
                ),
                2
            )

    }


# =============================================================================
# 30. PRIORITY MISSING SKILLS
# =============================================================================

def get_priority_missing_skills(

    result_df,

    max_jobs=1000

):

    counter = {}


    subset = (
        result_df
        .head(max_jobs)
    )


    for skills in (

        subset[
            "missing_skills"
        ]

    ):

        for skill in skills:

            if skill_weight(
                skill
            ) <= 0:

                continue


            counter[
                skill
            ] = (

                counter.get(
                    skill,
                    0
                )

                +

                1

            )


    ranked = sorted(

        counter.items(),

        key=lambda x:
            x[1],

        reverse=True

    )


    return [

        skill

        for skill, count
        in ranked[:15]

    ]


# =============================================================================
# 31. WHAT-IF ANALYSIS
# =============================================================================

def what_if_analysis(

    current_skills,

    added_skills,

    candidate_experience,

    preferred_city=None,

    preferred_work_mode=None,

    target_role=None

):

    before_results = (
        analyze_candidate(

            current_skills,

            candidate_experience,

            preferred_city,

            preferred_work_mode,

            target_role

        )
    )


    after_skills = (

        list(
            current_skills
        )

        +

        list(
            added_skills
        )

    )


    after_results = (
        analyze_candidate(

            after_skills,

            candidate_experience,

            preferred_city,

            preferred_work_mode,

            target_role

        )
    )


    before = (
        create_candidate_dashboard(
            before_results
        )
    )


    after = (
        create_candidate_dashboard(
            after_results
        )
    )


    return {

        "before_results":
            before_results,

        "after_results":
            after_results,

        "before":
            before,

        "after":
            after,

        "additional_eligible":
            (

                after[
                    "eligible_jobs"
                ]

                -

                before[
                    "eligible_jobs"
                ]

            ),

        "score_improvement":
            round(

                after[
                    "average_match_score"
                ]

                -

                before[
                    "average_match_score"
                ],

                2

            )

    }


# =============================================================================
# 32. FIXED DEMO CANDIDATE
#
# NO USER INPUT
#
# This only tests the backend.
# =============================================================================

DEMO_CANDIDATE_SKILLS = [

    "Python",
    "SQL",
    "Excel",
    "Pandas"

]


DEMO_EXPERIENCE = 1.0


DEMO_CITY = "Mumbai"


DEMO_WORK_MODE = "Remote"


DEMO_TARGET_ROLE = "Data Analyst"


# =============================================================================
# 33. RUN DEMO
# =============================================================================

print("\n" + "=" * 80)
print("CANDIDATE BACKEND DEMONSTRATION")
print("=" * 80)


demo_results = analyze_candidate(

    candidate_skills=
        DEMO_CANDIDATE_SKILLS,

    candidate_experience=
        DEMO_EXPERIENCE,

    preferred_city=
        DEMO_CITY,

    preferred_work_mode=
        DEMO_WORK_MODE,

    target_role=
        DEMO_TARGET_ROLE

)


demo_dashboard = (
    create_candidate_dashboard(
        demo_results
    )
)


print(
    "Candidate skills:",
    DEMO_CANDIDATE_SKILLS
)

print(
    "Experience:",
    DEMO_EXPERIENCE,
    "years"
)

print(
    "Preferred city:",
    DEMO_CITY
)

print(
    "Preferred work mode:",
    DEMO_WORK_MODE
)

print(
    "Target role:",
    DEMO_TARGET_ROLE
)


print("\n" + "-" * 80)
print("CANDIDATE OPPORTUNITY DASHBOARD")
print("-" * 80)


print(
    "Jobs analyzed:",
    demo_dashboard[
        "jobs_analyzed"
    ]
)

print(
    "Eligible jobs:",
    demo_dashboard[
        "eligible_jobs"
    ]
)

print(
    "Not eligible jobs:",
    demo_dashboard[
        "not_eligible_jobs"
    ]
)

print(
    "Eligibility percentage:",
    demo_dashboard[
        "eligibility_percent"
    ],
    "%"
)

print(
    "Excellent Fit:",
    demo_dashboard[
        "excellent_fit"
    ]
)

print(
    "Good Fit:",
    demo_dashboard[
        "good_fit"
    ]
)

print(
    "Potential Fit:",
    demo_dashboard[
        "potential_fit"
    ]
)

print(
    "Low Fit:",
    demo_dashboard[
        "low_fit"
    ]
)

print(
    "Very Low Fit:",
    demo_dashboard[
        "very_low_fit"
    ]
)

print(
    "Average Match Score:",
    demo_dashboard[
        "average_match_score"
    ]
)


# =============================================================================
# 34. TOP MATCHING JOBS
# =============================================================================

print("\n" + "=" * 80)
print("TOP 20 MATCHING JOBS")
print("=" * 80)


display(

    demo_results[

        [

            "job_id",
            "job_title",
            "company",
            "role",
            "city",
            "work_mode",
            "eligible",
            "skill_match_percent",
            "experience_match_percent",
            "role_relevance_percent",
            "match_score",
            "fit_category",
            "application_priority",
            "salary_lpa"

        ]

    ]

    .head(20)

)


# =============================================================================
# 35. EXPLAINABILITY
# =============================================================================

print("\n" + "=" * 80)
print("WHY IS THE CANDIDATE ELIGIBLE / NOT ELIGIBLE?")
print("=" * 80)


for _, row in (
    demo_results
    .head(5)
    .iterrows()
):

    print(
        "\nJob:",
        row[
            "job_title"
        ]
    )

    print(
        "Company:",
        row[
            "company"
        ]
    )

    print(
        "Status:",
        (
            "ELIGIBLE"
            if row[
                "eligible"
            ]
            else
            "NOT ELIGIBLE"
        )
    )

    print(
        "Match Score:",
        row[
            "match_score"
        ],
        "%"
    )

    print(
        "Skill Match:",
        row[
            "skill_match_percent"
        ],
        "%"
    )

    print(
        "Experience Match:",
        row[
            "experience_match_percent"
        ],
        "%"
    )

    print(
        "Role Relevance:",
        row[
            "role_relevance_percent"
        ],
        "%"
    )

    print(
        "Matched Skills:",
        ", ".join(
            row[
                "matched_skills"
            ]
        )
        if row[
            "matched_skills"
        ]
        else
        "None"
    )

    print(
        "Missing Skills:",
        ", ".join(
            row[
                "missing_skills"
            ]
        )
        if row[
            "missing_skills"
        ]
        else
        "None"
    )

    print(
        "Explanation:",
        row[
            "eligibility_explanation"
        ]
    )


# =============================================================================
# 36. PRIORITY SKILL GAPS
# =============================================================================

priority_missing_skills = (
    get_priority_missing_skills(
        demo_results
    )
)


print("\n" + "=" * 80)
print("PRIORITY SKILL GAPS")
print("=" * 80)


for i, skill in enumerate(

    priority_missing_skills[:10],

    start=1

):

    print(
        f"{i}. {skill}"
    )


# =============================================================================
# 37. WHAT-IF TEST
# =============================================================================

DEMO_ADDED_SKILLS = [

    "Power BI",
    "Data Analysis"

]


print("\n" + "=" * 80)
print("WHAT-IF ANALYSIS")
print("=" * 80)


demo_what_if = (
    what_if_analysis(

        current_skills=
            DEMO_CANDIDATE_SKILLS,

        added_skills=
            DEMO_ADDED_SKILLS,

        candidate_experience=
            DEMO_EXPERIENCE,

        preferred_city=
            DEMO_CITY,

        preferred_work_mode=
            DEMO_WORK_MODE,

        target_role=
            DEMO_TARGET_ROLE

    )
)


print(
    "Added skills:",
    DEMO_ADDED_SKILLS
)


print(
    "\nBEFORE:"
)

print(
    "Eligible jobs:",
    demo_what_if[
        "before"
    ][
        "eligible_jobs"
    ]
)

print(
    "Average Match Score:",
    demo_what_if[
        "before"
    ][
        "average_match_score"
    ]
)


print(
    "\nAFTER:"
)

print(
    "Eligible jobs:",
    demo_what_if[
        "after"
    ][
        "eligible_jobs"
    ]
)

print(
    "Average Match Score:",
    demo_what_if[
        "after"
    ][
        "average_match_score"
    ]
)


print(
    "\nAdditional eligible opportunities:",
    demo_what_if[
        "additional_eligible"
    ]
)


print(
    "Match Score improvement:",
    demo_what_if[
        "score_improvement"
    ]
)


# -------------------------------------------------------------------------
# What-if visualizations
# -------------------------------------------------------------------------

plt.figure(
    figsize=(8, 5)
)


plt.bar(

    [
        "Before",
        "After"

    ],

    [

        demo_what_if[
            "before"
        ][
            "average_match_score"
        ],

        demo_what_if[
            "after"
        ][
            "average_match_score"
        ]

    ]

)


plt.ylim(
    0,
    100
)


plt.title(
    "Average Match Score Before vs After Skill Addition"
)

plt.ylabel(
    "Average Match Score"
)

plt.tight_layout()

plt.show()


plt.figure(
    figsize=(8, 5)
)


plt.bar(

    [
        "Before",
        "After"

    ],

    [

        demo_what_if[
            "before"
        ][
            "eligible_jobs"
        ],

        demo_what_if[
            "after"
        ][
            "eligible_jobs"
        ]

    ]

)


plt.title(
    "Eligible Jobs Before vs After Skill Addition"
)

plt.ylabel(
    "Number of Eligible Jobs"
)

plt.tight_layout()

plt.show()


# =============================================================================
# 38. LANGCHAIN + GEMINI
#
# THIS SECTION WILL ASK FOR YOUR GEMINI API KEY.
# =============================================================================

print("\n" + "=" * 80)
print("LANGCHAIN + GEMINI")
print("=" * 80)


from getpass import getpass


print(
    """
You will now be asked for your Gemini API key.

Get/copy it from:
Google AI Studio -> API Keys

The key will be entered securely and will not
be displayed on screen.
"""
)


# ALWAYS ASK FOR THE KEY
GOOGLE_API_KEY = getpass(
    "Paste your Gemini API key here: "
)


if not GOOGLE_API_KEY.strip():

    raise ValueError(
        "Gemini API key was empty."
    )


os.environ[
    "GOOGLE_API_KEY"
] = GOOGLE_API_KEY.strip()


# =============================================================================
# 39. CREATE GEMINI MODEL
# =============================================================================

from langchain_google_genai import (
    ChatGoogleGenerativeAI
)

from langchain.tools import tool

from langchain.agents import (
    create_agent
)


# Use a broadly supported Gemini model.
GEMINI_MODEL = (
    "gemini-3.6-flash"
)


llm = ChatGoogleGenerativeAI(

    model=
        GEMINI_MODEL,

    temperature=
        0,

    max_retries=
        2

)


print(
    "\nTesting Gemini connection..."
)


try:

    test_response = llm.invoke(
        "Reply with exactly: Gemini connection successful."
    )


    print(
        "Gemini connection successful."
    )


except Exception as error:

    print(
        "\nGemini connection failed."
    )

    print(
        "Error:",
        error
    )

    print(
        """
Check:
1. API key is correct.
2. Gemini API is enabled for the project.
3. The key has not been exposed/revoked.
4. Your Google AI Studio project has API access.
"""
    )

    raise


# =============================================================================
# 40. LANGCHAIN TOOLS
# =============================================================================


@tool
def market_overview_tool():
    """
    Return actual market statistics calculated from the dataset.
    """

    return {

        "total_jobs":
            int(
                len(df)
            ),

        "unique_companies":
            int(
                df[
                    "company_name"
                ].nunique()
            ),

        "unique_cities":
            int(
                df[
                    "primary_city"
                ].nunique()
            ),

        "role_distribution":
            df[
                "role_category"
            ]
            .value_counts()
            .to_dict(),

        "top_skills":
            top_skills
            .head(15)
            .to_dict(),

        "fresher_friendly_percent":
            round(
                float(
                    df[
                        "is_fresher_friendly"
                    ].mean()
                    *
                    100
                ),
                2
            ),

        "remote_hybrid_percent":
            round(
                float(
                    df[
                        "remote_or_hybrid"
                    ].mean()
                    *
                    100
                ),
                2
            )

    }


@tool
def role_analysis_tool(
    role: str
):
    """
    Return role demand, salary, fresher accessibility,
    flexibility and project-defined opportunity score.
    """

    matches = [

        role_name

        for role_name
        in role_analysis.index

        if str(
            role_name
        ).lower()
        ==
        str(
            role
        ).lower()

    ]


    if not matches:

        return {

            "error":
                f"Role '{role}' not found."

        }


    actual_role = matches[0]


    row = (
        role_analysis
        .loc[
            actual_role
        ]
    )


    return {

        "role":
            actual_role,

        "job_count":
            int(
                row[
                    "job_count"
                ]
            ),

        "average_salary_lpa":
            round(
                float(
                    row[
                        "avg_salary"
                    ]
                ),
                2
            ),

        "fresher_rate_percent":
            round(
                float(
                    row[
                        "fresher_rate"
                    ]
                    *
                    100
                ),
                2
            ),

        "remote_hybrid_percent":
            round(
                float(
                    row[
                        "flexibility_rate"
                    ]
                    *
                    100
                ),
                2
            ),

        "opportunity_score":
            round(
                float(
                    row[
                        "opportunity_score_100"
                    ]
                ),
                2
            )

    }


@tool
def role_skills_tool(
    role: str
):
    """
    Return important core skills for a role.
    """

    result = (

        skill_long[

            (

                skill_long[
                    "role_category"
                ]
                .astype(str)
                .str.lower()
                ==
                str(
                    role
                ).lower()

            )

            &

            (

                skill_long[
                    "skill_weight"
                ]
                >
                0

            )

        ]

        .groupby(
            "skill"
        )[

            "job_id"

        ]

        .nunique()

        .sort_values(
            ascending=False
        )

        .head(15)

    )


    return {

        "role":
            role,

        "skills":
            result.to_dict()

    }


@tool
def candidate_analysis_tool(

    current_skills: str,

    experience: float,

    city: str = "",

    work_mode: str = "",

    target_role: str = ""

):
    """
    Analyze a candidate against the complete job dataset.
    """

    skills = [

        item.strip()

        for item
        in current_skills.split(",")

        if item.strip()

    ]


    results = analyze_candidate(

        candidate_skills=
            skills,

        candidate_experience=
            float(
                experience
            ),

        preferred_city=
            city
            if city.strip()
            else None,

        preferred_work_mode=
            work_mode
            if work_mode.strip()
            else None,

        target_role=
            target_role
            if target_role.strip()
            else None

    )


    summary = (
        create_candidate_dashboard(
            results
        )
    )


    missing = (
        get_priority_missing_skills(
            results
        )
    )


    return {

        "summary":
            summary,

        "priority_missing_skills":
            missing,

        "top_jobs":
            results[

                [

                    "job_id",
                    "job_title",
                    "company",
                    "role",
                    "city",
                    "work_mode",
                    "eligible",
                    "match_score",
                    "fit_category",
                    "application_priority"

                ]

            ]
            .head(10)
            .to_dict(
                orient="records"
            )

    }


@tool
def candidate_what_if_tool(

    current_skills: str,

    added_skills: str,

    experience: float,

    city: str = "",

    work_mode: str = "",

    target_role: str = ""

):
    """
    Perform before-vs-after candidate skill simulation.
    """

    current = [

        item.strip()

        for item
        in current_skills.split(",")

        if item.strip()

    ]


    added = [

        item.strip()

        for item
        in added_skills.split(",")

        if item.strip()

    ]


    result = what_if_analysis(

        current_skills=
            current,

        added_skills=
            added,

        candidate_experience=
            float(
                experience
            ),

        preferred_city=
            city
            if city.strip()
            else None,

        preferred_work_mode=
            work_mode
            if work_mode.strip()
            else None,

        target_role=
            target_role
            if target_role.strip()
            else None

    )


    return {

        "before":
            result[
                "before"
            ],

        "after":
            result[
                "after"
            ],

        "additional_eligible_jobs":
            result[
                "additional_eligible"
            ],

        "match_score_improvement":
            result[
                "score_improvement"
            ],

        "added_skills":
            added

    }


# =============================================================================
# 41. CREATE LANGCHAIN AGENT
# =============================================================================

langchain_tools = [

    market_overview_tool,

    role_analysis_tool,

    role_skills_tool,

    candidate_analysis_tool,

    candidate_what_if_tool

]


agent = create_agent(

    model=
        llm,

    tools=
        langchain_tools,

    system_prompt="""

You are an AI Job Market Analyst.

You analyze an Indian technology job-posting dataset.

Always use the available tools when numerical or
dataset-specific information is required.

Important rules:

1. Never invent statistics.

2. Match Score is a compatibility measure.
   It is NOT a probability of getting hired.

3. The dataset has no historical selected/rejected
   candidate outcomes.

4. Never claim that a candidate has an exact
   probability of being hired.

5. Salary statistics apply only to jobs with
   usable disclosed salary information.

6. Explain eligibility using the actual
   backend calculation.

7. Explain skill gaps clearly.

8. Use what-if analysis when the user wants
   to know the effect of learning a new skill.

9. Distinguish observed data from recommendations.

"""
)


print(
    "LangChain agent created successfully."
)


# =============================================================================
# 42. LANGCHAIN AUTOMATIC TEST
# =============================================================================

print("\n" + "=" * 80)
print("LANGCHAIN TEST")
print("=" * 80)


langchain_test_prompt = f"""

Analyze the following candidate:

Skills:
{", ".join(DEMO_CANDIDATE_SKILLS)}

Experience:
{DEMO_EXPERIENCE} years

City:
{DEMO_CITY}

Work mode:
{DEMO_WORK_MODE}

Target role:
{DEMO_TARGET_ROLE}

Please explain:

1. Number of jobs analyzed
2. Number of eligible jobs
3. Average match score
4. Best matching opportunities
5. Important skill gaps
6. Skills the candidate should learn next
7. What-if effect of adding Power BI and Data Analysis

Remember:
Do not describe Match Score as hiring probability.
"""


try:

    response = agent.invoke({

        "messages": [

            {

                "role":
                    "user",

                "content":
                    langchain_test_prompt

            }

        ]

    })


    final_message = (
        response[
            "messages"
        ][-1]
    )


    try:

        ai_text = (
            final_message.text
        )

    except Exception:

        if isinstance(
            final_message.content,
            str
        ):

            ai_text = (
                final_message.content
            )

        else:

            ai_text = str(
                final_message.content
            )


    print(
        ai_text
    )


except Exception as error:

    print(
        "LangChain test failed:"
    )

    print(
        error
    )


# =============================================================================
# 43. LANGGRAPH STATE
# =============================================================================

print("\n" + "=" * 80)
print("LANGGRAPH")
print("=" * 80)


from langgraph.graph import (
    StateGraph,
    START,
    END
)


class CareerState(
    TypedDict,
    total=False
):

    candidate_skills: List[str]

    candidate_experience: float

    preferred_city: Optional[str]

    preferred_work_mode: Optional[str]

    target_role: Optional[str]

    jobs_analyzed: int

    eligible_jobs: int

    average_match_score: float

    top_jobs: List[
        Dict[str, Any]
    ]

    missing_skills: List[str]

    recommended_skills: List[str]

    what_if: Dict[str, Any]

    final_explanation: str


# =============================================================================
# 44. LANGGRAPH NODE 1
# =============================================================================

def graph_match_node(
    state: CareerState
):

    results = analyze_candidate(

        candidate_skills=
            state[
                "candidate_skills"
            ],

        candidate_experience=
            state[
                "candidate_experience"
            ],

        preferred_city=
            state.get(
                "preferred_city"
            ),

        preferred_work_mode=
            state.get(
                "preferred_work_mode"
            ),

        target_role=
            state.get(
                "target_role"
            )

    )


    summary = (
        create_candidate_dashboard(
            results
        )
    )


    top_jobs = (

        results[

            [

                "job_id",
                "job_title",
                "company",
                "role",
                "city",
                "work_mode",
                "eligible",
                "skill_match_percent",
                "match_score",
                "fit_category"

            ]

        ]

        .head(10)

        .to_dict(
            orient="records"
        )

    )


    missing_skills = (
        get_priority_missing_skills(
            results
        )
    )


    return {

        "jobs_analyzed":
            summary[
                "jobs_analyzed"
            ],

        "eligible_jobs":
            summary[
                "eligible_jobs"
            ],

        "average_match_score":
            summary[
                "average_match_score"
            ],

        "top_jobs":
            top_jobs,

        "missing_skills":
            missing_skills

    }


# =============================================================================
# 45. LANGGRAPH NODE 2
# =============================================================================

def graph_recommendation_node(
    state: CareerState
):

    recommendations = (

        state[
            "missing_skills"
        ][:3]

    )


    return {

        "recommended_skills":
            recommendations

    }


# =============================================================================
# 46. LANGGRAPH NODE 3
# =============================================================================

def graph_what_if_node(
    state: CareerState
):

    added = (
        state[
            "recommended_skills"
        ]
    )


    if not added:

        return {

            "what_if":
                {}

        }


    result = what_if_analysis(

        current_skills=
            state[
                "candidate_skills"
            ],

        added_skills=
            added,

        candidate_experience=
            state[
                "candidate_experience"
            ],

        preferred_city=
            state.get(
                "preferred_city"
            ),

        preferred_work_mode=
            state.get(
                "preferred_work_mode"
            ),

        target_role=
            state.get(
                "target_role"
            )

    )


    return {

        "what_if": {

            "added_skills":
                added,

            "before_eligible":
                result[
                    "before"
                ][
                    "eligible_jobs"
                ],

            "after_eligible":
                result[
                    "after"
                ][
                    "eligible_jobs"
                ],

            "additional_eligible":
                result[
                    "additional_eligible"
                ],

            "before_score":
                result[
                    "before"
                ][
                    "average_match_score"
                ],

            "after_score":
                result[
                    "after"
                ][
                    "average_match_score"
                ],

            "score_improvement":
                result[
                    "score_improvement"
                ]

        }

    }


# =============================================================================
# 47. LANGGRAPH NODE 4
# =============================================================================

def graph_explanation_node(
    state: CareerState
):

    prompt = f"""

You are the final AI job-market analyst.

Candidate:
{state["candidate_skills"]}

Experience:
{state["candidate_experience"]}

City:
{state.get("preferred_city")}

Work mode:
{state.get("preferred_work_mode")}

Target role:
{state.get("target_role")}

Jobs analyzed:
{state.get("jobs_analyzed")}

Eligible jobs:
{state.get("eligible_jobs")}

Average Match Score:
{state.get("average_match_score")}

Top jobs:
{state.get("top_jobs", [])}

Missing skills:
{state.get("missing_skills", [])}

Recommended skills:
{state.get("recommended_skills", [])}

What-if:
{state.get("what_if", {})}

Create a final structured report containing:

1. Eligibility Status
2. Candidate Match Score
3. Job Opportunity Summary
4. Main Skill Gaps
5. Recommended Skills
6. What-if Improvement
7. Final Recommendation

Important:
Match Score is NOT hiring probability.
Do not invent numbers.
Do not say the candidate is guaranteed to get hired.

"""


    try:

        response = (
            llm.invoke(
                prompt
            )
        )


        try:

            explanation = (
                response.text
            )

        except Exception:

            if isinstance(
                response.content,
                str
            ):

                explanation = (
                    response.content
                )

            else:

                explanation = str(
                    response.content
                )


    except Exception as error:

        explanation = (

            "AI explanation unavailable. "
            f"Backend results are still valid. Error: {error}"

        )


    return {

        "final_explanation":
            explanation

    }


# =============================================================================
# 48. BUILD LANGGRAPH
# =============================================================================

career_graph = StateGraph(
    CareerState
)


career_graph.add_node(
    "match",
    graph_match_node
)


career_graph.add_node(
    "recommendation",
    graph_recommendation_node
)


career_graph.add_node(
    "what_if",
    graph_what_if_node
)


career_graph.add_node(
    "explanation",
    graph_explanation_node
)


career_graph.add_edge(
    START,
    "match"
)


career_graph.add_edge(
    "match",
    "recommendation"
)


career_graph.add_edge(
    "recommendation",
    "what_if"
)


career_graph.add_edge(
    "what_if",
    "explanation"
)


career_graph.add_edge(
    "explanation",
    END
)


compiled_career_graph = (
    career_graph.compile()
)


print(
    "LangGraph compiled successfully."
)


# =============================================================================
# 49. RUN LANGGRAPH
# =============================================================================

print("\n" + "=" * 80)
print("LANGGRAPH END-TO-END TEST")
print("=" * 80)


graph_input = {

    "candidate_skills":
        DEMO_CANDIDATE_SKILLS,

    "candidate_experience":
        DEMO_EXPERIENCE,

    "preferred_city":
        DEMO_CITY,

    "preferred_work_mode":
        DEMO_WORK_MODE,

    "target_role":
        DEMO_TARGET_ROLE

}


graph_result = (
    compiled_career_graph
    .invoke(
        graph_input
    )
)


print(
    graph_result[
        "final_explanation"
    ]
)


# =============================================================================
# 50. GRAPH VISUALIZATION
# =============================================================================

try:

    from IPython.display import (
        Image,
        display
    )


    graph_image = (
        compiled_career_graph
        .get_graph()
        .draw_mermaid_png()
    )


    print(
        "\nLangGraph Workflow:"
    )


    display(
        Image(
            graph_image
        )
    )


except Exception as error:

    print(
        "\nGraph image rendering unavailable."
    )

    print(
        """
START
  ↓
Candidate-Job Matching
  ↓
Skill Gap
  ↓
Recommendation
  ↓
What-if
  ↓
AI Explanation
  ↓
END
"""
    )


# =============================================================================
# 51. SAVE BACKEND
# =============================================================================

print("\n" + "=" * 80)
print("SAVING BACKEND")
print("=" * 80)


BACKEND_FOLDER = Path(
    "job_market_backend"
)


BACKEND_FOLDER.mkdir(
    exist_ok=True
)


# Cleaned dataset
df.to_csv(

    BACKEND_FOLDER
    /
    "cleaned_job_market_data.csv",

    index=False

)


# Skill dataset
skill_long.to_csv(

    BACKEND_FOLDER
    /
    "job_skills_long_format.csv",

    index=False

)


# Role opportunity scores
role_analysis.reset_index().to_csv(

    BACKEND_FOLDER
    /
    "role_opportunity_scores.csv",

    index=False

)


# Top skills
top_skills.reset_index(
    names="skill"
).to_csv(

    BACKEND_FOLDER
    /
    "top_skills.csv",

    index=False

)


# Job skill cache
joblib.dump(

    job_skill_cache,

    BACKEND_FOLDER
    /
    "job_skill_cache.pkl"

)


# Demo matching results
demo_results.to_csv(

    BACKEND_FOLDER
    /
    "demo_candidate_job_results.csv",

    index=False

)


# Demo summary
with open(

    BACKEND_FOLDER
    /
    "demo_candidate_summary.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        demo_dashboard,

        file,

        indent=4

    )


# What-if summary
with open(

    BACKEND_FOLDER
    /
    "demo_what_if_summary.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        {

            "added_skills":
                DEMO_ADDED_SKILLS,

            "before":
                demo_what_if[
                    "before"
                ],

            "after":
                demo_what_if[
                    "after"
                ],

            "additional_eligible":
                demo_what_if[
                    "additional_eligible"
                ],

            "score_improvement":
                demo_what_if[
                    "score_improvement"
                ]

        },

        file,

        indent=4

    )


# Skill taxonomy
taxonomy = {

    "technical_skills":
        sorted(
            technical_skills
        ),

    "business_skills":
        sorted(
            business_skills
        ),

    "generic_skills":
        sorted(
            generic_skills
        ),

    "weights": {

        "technical":
            1.0,

        "business":
            0.5,

        "generic":
            0.0,

        "other":
            0.0

    }

}


with open(

    BACKEND_FOLDER
    /
    "skill_taxonomy.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        taxonomy,

        file,

        indent=4

    )


# Backend configuration
backend_config = {

    "project":
        "AI-Assisted Indian Tech Job Market Analysis 2026",

    "dataset_rows":
        int(
            len(df)
        ),

    "candidate_matching":
        "All job postings",

    "skill_score_weight":
        0.50,

    "experience_score_weight":
        0.25,

    "role_relevance_weight":
        0.10,

    "location_score_weight":
        0.10,

    "work_mode_score_weight":
        0.05,

    "minimum_skill_match":
        50,

    "eligibility_rule":
        "Minimum experience satisfied + 50% core skill match + relevant job title",

    "langchain":
        True,

    "langgraph":
        True,

    "selection_probability":
        False,

    "selection_probability_reason":
        (
            "No historical candidate selected/rejected outcomes "
            "exist in the dataset."
        ),

    "demo_candidate":
        {

            "skills":
                DEMO_CANDIDATE_SKILLS,

            "experience":
                DEMO_EXPERIENCE,

            "city":
                DEMO_CITY,

            "work_mode":
                DEMO_WORK_MODE,

            "target_role":
                DEMO_TARGET_ROLE

        }

}


with open(

    BACKEND_FOLDER
    /
    "backend_config.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        backend_config,

        file,

        indent=4

    )


# =============================================================================
# 52. FINAL STATUS
# =============================================================================

print("\n" + "=" * 80)
print("PROJECT BACKEND READY")
print("=" * 80)


print(
    f"\nAll {len(df):,} job postings are available to the matching engine."
)


print(
    "\nCompleted components:"
)


completed_components = [

    "CSV upload",

    "Data quality analysis",

    "Data cleaning",

    "Experience normalization",

    "Market visualization",

    "Salary analysis",

    "Skill extraction",

    "Skill taxonomy",

    "Generic skill filtering",

    "Role opportunity scoring",

    "Candidate-job matching",

    "Eligibility detection",

    "Explainable eligibility",

    "Candidate opportunity dashboard",

    "Top matching jobs",

    "Skill-gap analysis",

    "What-if analysis",

    "LangChain + Gemini",

    "LangGraph workflow",

    "Backend export"

]


for i, item in enumerate(

    completed_components,

    start=1

):

    print(
        f"{i}. {item}"
    )


print(
    "\nBackend saved at:"
)

print(
    BACKEND_FOLDER.resolve()
)


print(
    "\nGenerated files:"
)


for file in sorted(
    BACKEND_FOLDER.iterdir()
):

    print(
        "-",
        file.name
    )


print("\n" + "=" * 80)

print(
"""
FINAL ARCHITECTURE

23,201 JOBS
     ↓
DATA CLEANING
     ↓
SKILL TAXONOMY
     ↓
MARKET ANALYSIS
     ↓
CANDIDATE PROFILE
     ↓
CANDIDATE-JOB MATCHING
     ↓
ELIGIBILITY
     ↓
MATCH SCORE
     ↓
JOB RANKING
     ↓
SKILL GAP
     ↓
RECOMMENDATIONS
     ↓
WHAT-IF SIMULATION
     ↓
LANGCHAIN
     ↓
LANGGRAPH
     ↓
FINAL AI EXPLANATION
     ↓
FUTURE STREAMLIT UI

IMPORTANT:
The Match Score represents compatibility.
It is NOT the probability of being hired.

A genuine selection-probability model requires
historical selected/rejected candidate data.
"""
)

print(
    "=" * 80
)
```
